```python
# PROJECT AIM:
# lists of 5 most popular baby names in 2014 in US... 
# according to the statistical records on the CSV file?
import pandas as pd
```

```python
us_baby_names = pd.read_csv('012 us_baby_names.csv')
us_baby_names
```

```python
us_baby_names2014 = us_baby_names.loc[us_baby_names['Year'] == 2014, :]
us_baby_names2014
```

```python
sorted_us_baby_names2014 = us_baby_names2014.sort_values('Count', ascending=True)
sorted_us_baby_names2014
```

```python
#These were the list of five USA babys' names that are not popular in the year 2014
five_most_uncommon_us_baby_names2014 = sorted_us_baby_names2014.iloc[0:5]
five_most_uncommon_us_baby_names2014
```

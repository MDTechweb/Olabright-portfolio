```python
import pandas as pd
```

```python
crime = pd.read_csv('crime_boston.csv', encoding='latin-1')
```

```python
crime
```

```python
# The isnull() method ask if there is a specific space in the data that is empty and return a boolean True if it is empty...
# and False if it is otherwise
crime.isnull()
```

```python
# This check each row and ask if there is any empty spaces in that particular row and return "True" 
# for the particular row if it sees one and "False" if it sees non
crime.isnull().any(axis=1)
```

```python
# This list out every row with empty values or empty spaces
rows_with_missing_values = crime.isnull().any(axis=1)
crime[rows_with_missing_values]
```

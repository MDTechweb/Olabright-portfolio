```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn
```

```python
listings = pd.read_csv('listings.csv')
```

```python
listings
```

```python
# The "hue" tells the "seaborn" library to add multi-color to the bars
sn.countplot(x = 'neighbourhood_group', data = listings, hue = 'neighbourhood_group')
plt.xlabel('Neighbourhood groups listing')
plt.ylabel('Total numbeer of listings')
plt.show()
```

```python
# This takes the "neighbourhood_goup" as the x axis and "price" as the y axis and performs an average operation in the total..
# price listing for each "neighbourhood_group" and give a total average of the price for each "neighbourhood_group" listing
# Also, the longer the black thin line, the more uncertainty but the shorter it is, the more certainly you can trust the price ranges around
# the blue bar

# The "hue" tells the "seaborn" library to add multi-color to the bars
sn.barplot(x = 'neighbourhood_group', y = 'price', data = listings, hue = 'neighbourhood_group')
plt.show()
```

```python
# This typically removes the black thin line, but it is still the same thing it does with the previous average barplot!
sn.barplot(x = 'neighbourhood_group', y = 'price', data = listings, hue = 'neighbourhood_group', errorbar = ('ci', False))
plt.show()
```
